```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import matplotlib
import os
import numpy as np
from PIL import Image


path_images_dir = "capture_images"

image_files = sorted(
    f for f in os.listdir(path_images_dir)
    if f.lower().endswith(".jpg")
)

if not image_files:
    raise ValueError("No JPG images found")

images = [
    np.array(Image.open(os.path.join(path_images_dir, f)))
    for f in image_files
]

fig, ax = plt.subplots()
ax.axis("off")

im = ax.imshow(images[0])

def update(frame):
    im.set_data(images[frame])
    return im,

ani = animation.FuncAnimation(
    fig,
    update,
    frames=len(images),
    interval=50,
    blit=False
)

ani.save(
    "movie.gif",
    writer=animation.PillowWriter(fps=20)
)

plt.show()
```
